```python
import pandas as pd

print("CodeOrbit Sales Analysis")
print("Pandas is ready!")
```

```python
sales = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Pen", "Notebook", "Pen", "Bag", "Notebook", "Bag"],
    "Category": ["Stationery", "Stationery", "Stationery",
                 "Accessories", "Stationery", "Accessories"],
    "Region": ["South", "North", "South", "East", "North", "East"],
    "Quantity": [3, 2, 5, 1, 4, 2],
    "Price": [10, 50, 10, 500, 50, 500]
})

print(sales)
```

```python
sales["Revenue"] = sales["Quantity"] * sales["Price"]

print(sales[["Product", "Quantity", "Price", "Revenue"]])
print("Total Revenue:", sales["Revenue"].sum())
```

```python
average_order_value = sales["Revenue"].mean()

print("Average Order Value:", average_order_value)
```

```python
top_products = sales.groupby("Product")["Quantity"].sum()

print(top_products.sort_values(ascending=False))
```

```python
category_sales = sales.groupby("Category")["Revenue"].sum()

print(category_sales.sort_values(ascending=False))
```

```python
region_sales = sales.groupby("Region")["Revenue"].sum()

print(region_sales.sort_values(ascending=False))
```

```python
average_order_value = sales["Revenue"].sum() / sales["Order_ID"].nunique()

print("Average Order Value:", round(average_order_value, 2))
```

```python
sales["Order_Date"] = pd.to_datetime([
    "2026-01-05",
    "2026-01-12",
    "2026-02-03",
    "2026-02-15",
    "2026-03-08",
    "2026-03-20"
])

sales["Month"] = sales["Order_Date"].dt.strftime("%B")

monthly_sales = sales.groupby("Month")["Revenue"].sum()

print(monthly_sales)
```

```python
monthly_summary = sales.groupby("Month", sort=False).agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "nunique")
)

print(monthly_summary)
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Average Order Value",
        "Best-Selling Product",
        "Top Revenue Category",
        "Top Revenue Region"
    ],
    "Result": [
        sales["Revenue"].sum(),
        round(sales["Revenue"].sum() / sales["Order_ID"].nunique(), 2),
        sales.groupby("Product")["Quantity"].sum().idxmax(),
        sales.groupby("Category")["Revenue"].sum().idxmax(),
        sales.groupby("Region")["Revenue"].sum().idxmax()
    ]
})

print(summary)
```
